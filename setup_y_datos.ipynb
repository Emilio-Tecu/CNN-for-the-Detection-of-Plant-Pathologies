```python
# Ruta donde se guardarán las imagenes de otras personas.
path_otros = '/content/drive/MyDrive/Redes neuronales/Tarea 7. Reconocimiento facial/data/binary/otros'
os.makedirs(path_otros, exist_ok=True)

print("Descargando rostros para la clase negativa...")
# Descargamos 50 rostros aleatorios para equilibrar mis fotos.
lfw_people = fetch_lfw_people(min_faces_per_person=1, resize=None, slice_=None)
indices = np.random.choice(len(lfw_people.images), 50, replace=False)

for i, idx in enumerate(indices):
    img = lfw_people.images[idx]
    # Normalizar a 0-255 y formato uint8 para guardar como JPG.
    if img.max() <= 1.0: img = (img * 255).astype(np.uint8)
    else: img = img.astype(np.uint8)

    cv2.imwrite(os.path.join(path_otros, f'otro_{i}.jpg'), img)

print(f"Generadas 50 imágenes en {path_otros}")
```
